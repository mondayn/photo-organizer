```python
def ensure_lst(i):
    if not isinstance(i,list):
        return [i]
    return i

def label(pics:list):
    pics = ensure_lst(pics)
    return [
        get_encodings(i)
        [0] # get one & only fac in pic
        for i in pics
    ]

wife = label('labeled_pics/wife.JPG')
print(f'each face encodes to array of {len(wife[0])} length ')
```

```python
daughter = label([
        'labeled_pics/daughter.JPG',  
        'labeled_pics/daughter1.JPG',
])
save_encodings("labeled_embeddings/daughter.npz",daughter)


me = label([
    'labeled_pics/me.JPG'
])
save_encodings("labeled_embeddings/me.npz",me)
```

```python
def face_matches(known_faces,faces,tolerance=.6):

    known_faces=ensure_lst(known_faces)
    faces=ensure_lst(faces)
    
    for f in faces:
        # print('face shape',f.shape)
        for dist in np.linalg.norm(known_faces - f, axis=1):
            print('dist ',dist)
            if dist < tolerance: return True
    return False

print('does daughter at 18 months match 3 years? ',face_matches(daughter[0],daughter[1]),'\n\n')

print('daughter match wife? ',face_matches(daughter,wife),'\n\n')

print('me match wife? ',face_matches(me,wife))

```

```python
files = list(
    pd.read_parquet('features/unique.parquet').sample(5).path
)

def name_of_object(arg):
    for name, value in globals().items():
        if value is arg and not name.startswith('_'):
            return name
        
def all_matches(f):
    matches=[]
    for m in [wife,daughter,me]:
        if face_matches(m,get_encodings(f)):
            matches.append(name_of_object(m))
    return matches

for f in files:
    # show(f)
    print('matches are ',all_matches(f))
    # print(f)
    # print(f'"{f}" wife: {face_matches(wife,get_encodings(f))}, daughter:{face_matches(daughter,get_encodings(f))}')
```
